```python
print(">>===----银行管理系统----===<<")
class people:
    def __init__(self,name,qian=0):
        self.qian=qian
        self.name=name  
    def money(self,m):
        self.qian+=m
        print(f"存入{m}元，您的账户余额是{self.qian}元")
    def person(self):
        print(f"{self.name}的余额为{self.qian}元")
    def chu(self,w):
        if w<=0:
            print("没有余额")
            return 0
        if w>self.qian:
            print("余额不足")
            return 0
        self.qian-=w
        print(f"您已取走{w}元,您的余额为{self.qian}元")
p=people("蛙",10000000000)
while True:
    print("1取款 2存款 3查款")
    wx=input()
    if wx=="1":
        w=float(input("取出金额："))
        p.chu(w)
    elif wx=="2":
        m=float(input("存入金额："))
        p.money(m)
    elif wx=="3":
        p.person()
    elif wx=="0":
        break
    elif wx=="-1":
        print("谢谢使用")
        break
    else :
        print("请重新输入")
   
    


```
